# Ridge Regression


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.linear_model import RidgeCV
from sklearn.metrics import mean_squared_error, r2_score
from mlxtend.evaluate import bias_variance_decomp
import matplotlib.pyplot as plt
```

```python
# 0. Define Features (X) and Target (y)
# We use 'log_price' as the target as established in your preprocessing
X = data_frame.drop(columns=['log_price'])
y = data_frame['log_price']

# 1. Scale the features
# Ridge regression is sensitive to the scale of input features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Define a range of alphas to test
alphas = np.logspace(-3, 3, 100)

# 3. Initialize RidgeCV
ridge_cv = RidgeCV(alphas=alphas, scoring='neg_mean_squared_error', cv=5)

# 4. Fit the model on scaled training data
ridge_cv.fit(X_scaled, y)

# 5. Lamada Results
print(f"Best Alpha: {ridge_cv.alpha_}")
```

# Predict

```python
X_test = test_data_frame.drop(columns=['id', 'zipcode', 'date'])
# X_test = test_data_frame.drop(columns=['id', 'zipcode', 'date', 'price'])
X_test_scaled = scaler.transform(X_test)
log_predictions = ridge_cv.predict(X_test_scaled)
prices_predictions = np.exp(log_predictions)
test_data_frame['price'] = prices_predictions
print(test_data_frame[['id', 'price']].head())

submission = test_data_frame[['id', 'price']]
submission.to_csv('midterm_prediction_result.csv', index=False)
```

# Learning in the Project

```python
# 1. Define Features (X) and Target (y)
# We use 'log_price' as the target as established in your preprocessing
l_X = data_frame.drop(columns=['log_price'])
l_y = data_frame['log_price']

# 2. Split the data into training and testing sets
l_X_train, l_X_test, l_y_train, l_y_test = train_test_split(l_X, l_y, test_size=0.3, random_state=42)

# 3. Scale the features
# Ridge regression is sensitive to the scale of input features
scaler = StandardScaler()
l_X_train_scaled = scaler.fit_transform(l_X_train)
l_X_test_scaled = scaler.transform(l_X_test)
```

```python
# 4. pick different alpha
n_alphas = 50
alphas = np.logspace(-8, 6, n_alphas)

# record the MSE, BIAS, VAR
mse_list, bias_list, var_list = [], [], []
coefs = []
train_r2_scores = []
test_r2_scores = []

for a in alphas:
    # 5. Initialize and train the Ridge model
    model = Ridge(alpha=a, fit_intercept=True)

    # 6. MSE, Bias, Variance
    avg_expected_loss, avg_bias, avg_var = bias_variance_decomp(
        model, l_X_train_scaled, l_y_train.values, l_X_test_scaled, l_y_test.values,
        loss='mse', num_rounds=50, random_seed=42
    )
    mse_list.append(avg_expected_loss)
    bias_list.append(avg_bias**2) # Bias^2
    var_list.append(avg_var)

    # 7. coefficient
    model.fit(l_X_train_scaled, l_y_train)
    coefs.append(model.coef_)

    # 8. Make predictions
    l_y_train_pred = model.predict(l_X_train_scaled)
    l_y_test_pred = model.predict(l_X_test_scaled)
    train_r2 = r2_score(l_y_train, l_y_train_pred)
    test_r2 = r2_score(l_y_test, l_y_test_pred)
    train_r2_scores.append(train_r2)
    test_r2_scores.append(test_r2)
```

```python
alphas
```

```python
# Bias-Variance
plt.figure(figsize=(8, 3))
plt.plot(alphas, mse_list, label='Expected Test MSE', color='red', linewidth=2)
plt.plot(alphas, bias_list, label='Bias^2', color='blue', linestyle='--')
plt.plot(alphas, var_list, label='Variance', color='green', linestyle='--')
plt.xscale('log')
plt.xlabel('Alpha (Regularization Strength)')
plt.ylabel('Error')
plt.title('Ridge Regression: Bias-Variance Trade-off')
plt.legend()
plt.show()
```

The larger alpha is, the larger Bias we get, and the smaller Variance we get.

```python
# features
ax = plt.gca()
ax.plot(alphas, coefs)
ax.set_xscale("log")
ax.set_xlim(ax.get_xlim()[::-1])  # reverse axis
plt.xlabel("alpha")
plt.ylabel("weights")
plt.title("Ridge Coefficients vs Regularization Strength (alpha)")
plt.axis("tight")
# plt.legend(
#     [f"Feature {i + 1}" for i in range(X.shape[1])], loc="best", fontsize="small"
# )
plt.legend(X.columns, loc="best", fontsize="small", bbox_to_anchor=(1.05, 1))
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.plot(alphas, train_r2_scores, label='Train R^2', color='red', linewidth=2)
plt.plot(alphas, test_r2_scores, label='Test R^2', color='blue', linewidth=2)
plt.xscale('log')
plt.xlabel('Alpha (Regularization Strength)')
plt.ylabel('R^2')
plt.title('R^2 Changes')
plt.legend()
plt.show()
```

The larger alpha is, the less model fits both training and testing data.
